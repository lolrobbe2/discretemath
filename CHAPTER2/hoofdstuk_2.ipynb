### aantal_schrikkeljaren

Schrijf de functie `aantal_schrikkeljaren(jaartal: int) -> int`, die het aantal schrikkeljaren teruggeeft van het jaar 1 tot en met het jaar `jaartal`.

Gebruik geen lussen of andere ingewikkelde constructies!

**Voorbeeld:**

```console?lang=python&prompt=>>>
>>> aantal_schrikkeljaren(2025)
491
```

````python
def aantal_schrikkeljaren(jaartal: int) -> int:
    return (jaartal // 4) - (jaartal // 100) + (jaartal // 400)

print(aantal_schrikkeljaren(2029))
````

## Oefening 2.16
Ga voor volgende relaties na of het al dan niet (totale of partiële) functies, bijecties, injecties of surjecties zijn. Probeer zo specifiek mogelijk te zijn. Gebruik hierbij de functie `def genereer_grafiek(x: list, y: list, scatter: bool) -> None`, waarvan de werking hieronder geïllustreerd wordt.

````python
def genereer_grafiek(x: list, y: list, scatter: bool) -> None:
    fig = plt.figure()
    ax = fig.add_axes([0, 0, 1, 1])
    ax.spines["left"].set_position('zero')
    ax.spines["bottom"].set_position('zero')
    ax.spines["right"].set_color('none')
    ax.spines["top"].set_color('none')
    if scatter:
        plt.scatter(x, y)
    else:
        plt.plot(x, y)
    plt.xlabel("x")
    plt.ylabel("y", rotation=0)
    plt.show()
````
